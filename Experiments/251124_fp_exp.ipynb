```python
import pandas as pd

import fpgrowth as fp
import refinements as rf
```

```python
sessID = [ 'S1', 'S2', 'S3', 'S4', 'S5']
skills = [ ['E', 'K', 'M', 'N', 'O', 'Y'], 
           ['D', 'E', 'K', 'N', 'O', 'Y'],
           ['A', 'E', 'K', 'M'],
           ['K', 'M', 'Y'],
           ['C', 'E', 'I', 'K', 'O', 'O']]
itemType = [1, 2, 3, 4, 1]

df = pd.DataFrame( {'sessionID': sessID,
                    'skills': skills,
                    'itemType': itemType} )
df
```

```python
frequent_itemsets = fp.fpgrowth(subgroup=df, attribute_of_interest='skills', minimal_support=3, maximum_support=10, maximum_length=5)


for i in frequent_itemsets:
    print(i, frequent_itemsets[i])
```

```python
candidates = rf.nominal_refinement(subgroup=df, attribute='itemType') #, minimal_support=0.6, maximum_support=0.9, maximum_length=3)

for ca in candidates:
    print(ca)
```
