```python
import pandas as pd

import fpgrowth as fp
```

```python
sessID = [ 'S1', 'S2', 'S3', 'S4', 'S5']
skills = [ ['E', 'K', 'M', 'N', 'O', 'Y'], 
           ['D', 'E', 'K', 'N', 'O', 'Y'],
           ['A', 'E', 'K', 'M'],
           ['K', 'M', 'Y'],
           ['C', 'E', 'I', 'K', 'O', 'O']]

df = pd.DataFrame( {'sessionID': sessID,
                    'skills': skills} )
df
```

```python
frequent_itemsets = fp.fpgrowth(descriptors=df, attribute_of_interest='skills', ids=['sessionID'], minimal_support=3, maximum_support=10, maximum_length=5)


for i in frequent_itemsets:
    print(i, frequent_itemsets[i])
```
