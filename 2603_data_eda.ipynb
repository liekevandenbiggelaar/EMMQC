```python
def cat_hints(value):

    if value < 5:
        return str(value)
    
    elif value >= 5 and value < 10:
        return '5 - 10'
    
    elif value >= 10:
        return '10+'


def cat_steps(value):

    if value < 5:
        return str(value)
    
    elif value >= 5 and value < 10:
        return '5 - 10'
    
    elif value >= 10 and value < 20:
        return '10 - 20'
    
    elif value >= 20 and value < 30:
        return '20 - 30'

    elif value >= 30: # and this is already very much for steps
        return '30+'
```

```python
pd.set_option('display.max_columns', None)
df = pd.read_csv("algebrakit_practice_anonymized.csv")

df['hintsRequested_cat'] = [cat_hints( df.iloc[i]['hintsRequested']) for i in range(len(df))]

df['try_count'] = [cat_hints( df.iloc[i]['event_types_EVALUATE']) for i in range(len(df))]
df.head()
```

```python
pd.set_option("display.max_rows", None)
df['event_types_EVALUATE'].value_counts()
```

```python
df[['finished']].value_counts()[False] / ( df[['finished']].value_counts()[False] + df[['finished']].value_counts()[True])
```

```python
all = []

for ex in list(df['interactionID'].unique()):

    df_ex = df[df['interactionID'] == ex]
    alles = len(df_ex)
    passes = sum(list(df_ex['finished']))
    failes = alles - passes
    pass_rate = round(passes/alles, 2)
    fail_rate = round(failes/alles, 2)

    all.append( [ex, fail_rate, pass_rate])


all.sort(key=lambda x: x[1])

for k in all:
    print(f"{k[0]} has pass rate of {k[2]} and fail rate of {k[1]}")
```

```python
df_clean = pp.clean_data_algebrakit(unfilterd_data=df, filter_type="remove_interaction")
```

```python

descriptors = [ 'interactionType', 'primarySubject', 'secondarySubject', 'stepTypes', #categorical
                'skills', 'positive_skills', 'negative_skills', 'event_evaluations', # collections
                'context_hint', #boolean
                  'hintsRequested_cat', 'try_count']
                  #, 'stepTypes_ALGEBRA', 'stepTypes_CALCULATE', 'stepTypes_CONCEPT', 'stepTypes_STRATEGY']#, #ordinal
                #'hintsRequested'] #, 'event_types_HINT', 'event_types_EVALUATE',
                # 'event_evaluations_CORRECT', 'event_evaluations_ERROR',
                # 'event_evaluations_FALSE_EXPRESSION',
                # 'event_evaluations_HINT', 'event_evaluations_NO_MATCH',
                # 'event_evaluations_PARSE_ERROR', 'event_evaluations_TRUE_EXPRESSION',
                # 'event_evaluations_UNKNOWN_STEP'] #ordinal 

""" OPTIONAL WITH OR WITHOUT HINTSREQUESTED """

targets = ['finished'] # all to do with performance

ids = ['exerciseID', 'sessionID', 'questionID', 'interactionID', 'representative_index', 'studentName']

# +- 20 seconden (with naive collection this is 12 mins)
refinement_params = {
                     'numerical_bins': 5,
                     'minimal_support_collections': 0.05, 'maximum_support_collections': 0.95, 
                     'maximum_itemsets': None, 'closed_itemsets': True
                     }

beam_params = {'k': 50, 'd': 3, 'w': 25} 
constraint_params = {'minimum_size': 25, 'improvement_rate': 0.1}
model_params = {'model': 'F_beta', 'beta': 1, 'weighted': True, 'inner entropy levels': ['session'], 'outer entropy levels': ['interaction'],
                 'subgroup_level': 'session', 'target_level': 'interaction'}



result_set_st = store_bs.beam_search(data=df_clean, descriptors=descriptors, targets=targets, ids=ids, 
                            refinement_params=refinement_params, beam_params=beam_params,
                            constraint_params=constraint_params, model_params=model_params)

print(len(result_set_st._queue))
```

```python
examined = 1
selected = 1
for sub in result_set_st._queue:

    examined += 1

    fail_rate_ex = round(np.mean(sub._metrics['goodness_weights']), 2)
    fail_rate_sg = round(np.mean(sub._metrics['fail rates']), 2)
    rel_impr = round( (fail_rate_sg - fail_rate_ex) / fail_rate_ex, 2)

    if rel_impr >= 0.75:
        print(f"#{examined}: {sub}")
        selected += 1
        print(f" avg. recall = {np.mean(sub._metrics['recall'])}")
        print(f" avg. precision = {np.mean(sub._metrics['precision'])}")
        print(f" avg. F_beta = {round(np.mean(sub._metrics['F_beta']), 2)}")
        print(f" avg. goodness_weights = {round(np.mean(sub._metrics['goodness_weights']), 2)}")
        print(f" avg. fail rate = {round(np.mean(sub._metrics['fail rates']), 2)}")
        print(f" Relative improvement = {rel_impr}")
        print(f" sessions = {min(sub._metrics['session_counts']), max(sub._metrics['session_counts'])}")
        # print(f" inner entropy = {sub._metrics['inner entropy']}")
        # print(f"inner sg count {sub._metrics['n_subgroup_inner']}")
        # print(f"inner co count {sub._metrics['n_complement_inner']}")

        print(f" exercises = {sub._metrics['exercise_count']}")
        print(f" exercises = {sub._metrics['exercises']}")
        # print(f" outer entropy = {sub._metrics['outer entropy']}")
        # print(f"outer sg count {sub._metrics['n_subgroup_outer']}")
        # print(f"outer co count {sub._metrics['n_complement_outer']}")

        print(56 * "=========")

        # if selected == 25:
        #     break
```
